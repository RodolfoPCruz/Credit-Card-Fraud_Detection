# DataSet Description



The dataset was downloaded from Kaggle and is available at the following link: 

https://www.kaggle.com/datasets/mlg-ulb/creditcardfraud

It contains credit card transactions made in September 2013 by European cardholders. All transactions occurred over a two-day period. Due to confidentiality issues, neither the original features nor information about the data was provided. Features V1, V2, ..., V28 are the principal components obtained from applying PCA to the original features. Only two features were not transformed: Time and Amount. Time refers to the time elapsed between the current transaction and the first registered transaction, while amount is the transaction amount.

# Data Distribution

## Class

```python
class_count = data['Class'].value_counts()
class_count
```

```python
print('High imbalanced Dataset')
print(f'Only {class_count[1]} transactions out of {len(data)} were fraudulent.')
print(f"{100*(data['Class']==1).mean():.3f}% of the transactions were fraudulent")
```

```python
quantiles = data.loc[:,['Amount']].quantile(np.arange(0.1, 1, 0.1))
quantiles.index.name = 'quantiles'
quantiles
```
